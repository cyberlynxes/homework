```python
# Function to display the menu, validate input, and calculate total price
def order_pizza():
    print("=====================================")
    print("Welcome to Rita Li's Pizza Order System!")
    print("=====================================")

    # --- Data Definitions with Base Prices ---
    sizes = ["Small", "Medium", "Large"]
    size_prices = [8, 10, 12]  # Prices for each size

    types = ["Margherita", "Pepperoni", "Vegetarian", "BBQ Chicken"]
    type_prices = [2, 3, 2, 4]  # Additional cost for specific pizza types

    toppings = ["Olives", "Mushrooms", "Onions", "Peppers", "Extra Cheese"]
    topping_price_per_item = 1  # Cost per topping

    # --- Helper function for input validation ---
    # Ensures the user enters a valid number within the allowed range
    def get_valid_input(prompt_text, min_val, max_val):
        while True:
            try:
                input_val = int(input(prompt_text))
                if min_val <= input_val <= max_val:
                    return input_val
                else:
                    print(f"Invalid input. Please enter a number between {min_val} and {max_val}.")
            except ValueError:
                print("Invalid input. Please enter a valid number.")

    # --- 1. Get Size Choice ---
    print("\n--- Choose a pizza size ---")
    for i in range(len(sizes)):
        print(f"{i + 1} - {sizes[i]} (${size_prices[i]})")
    size_choice = get_valid_input("Enter the number corresponding to your choice: ", 1, len(sizes))

    # --- 2. Get Type Choice ---
    print("\n--- Choose a pizza type ---")
    for i in range(len(types)):
        print(f"{i + 1} - {types[i]} (${type_prices[i]})")
    type_choice = get_valid_input("Enter the number corresponding to your choice: ", 1, len(types))

    # --- 3. Get Toppings Choice ---
    print("\n--- Choose your toppings (Additional $1 each) ---")
    for i in range(len(toppings)):
        print(f"{i + 1} - {toppings[i]}")
    print("(e.g., type '1,3' for Olives and Onions, or press Enter for none)")

    topping_input = input("Enter your choices: ")
    selected_toppings = []  # Initialize empty list

    if topping_input.strip() != "":
        topping_choices = topping_input.split(",")
        # Filter out any invalid topping numbers to prevent errors
        for t in topping_choices:
            try:
                t_index = int(t.strip())
                if 1 <= t_index <= len(toppings):
                    selected_toppings.append(toppings[t_index - 1])
            except ValueError:
                continue

    # --- 4. Get Quantity ---
    quantity = get_valid_input("\nHow many of this pizza would you like to order? : ", 1, 10)

    # --- 5. Calculate Total Price ---
    # Formula: (Size Price + Type Price + (Number of Toppings * Topping Price)) * Quantity
    base_cost = size_prices[size_choice - 1] + type_prices[type_choice - 1] + (len(selected_toppings) * topping_price_per_item)
    total_bill = base_cost * quantity

    # --- 6. Print Order Summary ---
    print("\n=====================================")
    print("          ORDER SUMMARY")
    print("=====================================")
    print(f"Pizza Type : {types[type_choice - 1]}")
    print(f"Size       : {sizes[size_choice - 1]}")
    print(f"Toppings   : {', '.join(selected_toppings) if selected_toppings else 'None'}")
    print(f"Quantity   : {quantity}")
    print("-------------------------------------")
    print(f"Unit Price : ${base_cost}")
    print(f"TOTAL BILL : ${total_bill}")
    print("=====================================")

    print("\nThank you for your order at Rita Li's Pizza!")

# Call the function to start the order process
order_pizza()
```
